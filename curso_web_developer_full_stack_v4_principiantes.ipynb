# Módulo 0 — Mentalidad Full-Stack y mapa de la web

### A. Lo que vas a aprender
Vas a entender que una página web no es un archivo suelto: es un sistema con varias piezas que colaboran. Al terminar podrás dibujar y explicar ese sistema completo.

### B. ¿Qué es?
Cuando visitas una web, tu navegador le pide información a una computadora que está en algún lugar de internet (un servidor). Esa computadora procesa la petición, puede consultar datos guardados, y devuelve una respuesta que tu navegador convierte en la página que ves. "Full-stack" significa aprender las piezas principales de ese recorrido: la interfaz (frontend), el servidor que decide qué hacer (backend) y el lugar donde se guardan los datos (base de datos).

### C. ¿Por qué existe?
Si no entiendes el recorrido completo, cada tecnología nueva parece un bloque aislado y mágico. Entender el mapa completo desde el primer día te permite ubicar cada herramienta futura (HTML, CSS, Express, PostgreSQL, React...) dentro de un lugar concreto del sistema, en vez de memorizar nombres sueltos.

### D. ¿Cómo funciona? (paso a paso)
1. Escribes una dirección (URL) o haces clic en un botón.
2. El navegador arma una solicitud y la envía por internet.
3. Un servidor recibe la solicitud y decide qué hacer con ella.
4. Si hace falta, el servidor consulta o modifica una base de datos.
5. El servidor arma una respuesta (HTML, o datos en formato JSON).
6. El navegador recibe la respuesta y actualiza lo que ves en pantalla.

### E. Diagrama visual

```python
   USUARIO
      |
      v
  NAVEGADOR
      |
      | solicitud (HTTP)
      v
  SERVIDOR ------------> BASE DE DATOS
      |    <------------
      | respuesta (HTTP)
      v
  NAVEGADOR
      |
      v
   USUARIO
```

### F. Ejemplo cotidiano (analogía)
> Piensa en un restaurante. El frontend es el salón y el menú que ve el cliente. El backend es la cocina: recibe el pedido y decide qué hacer con él. La base de datos se parece a la despensa y los registros del restaurante. HTTP es el camino por el que viaja el pedido del salón a la cocina, y la respuesta de la cocina de vuelta al salón.

```python
Usuario -> Navegador -> HTTP -> Servidor
  -> Base de datos -> Servidor -> Navegador
```

### I. Explicación línea por línea

| Línea | Qué hace |
|---|---|
| `Usuario -> Navegador` | La persona interactúa con la interfaz: hace clic, escribe, navega. |
| `Navegador -> HTTP` | El navegador convierte esa acción en una solicitud con un formato estándar. |
| `HTTP -> Servidor` | La solicitud viaja por internet hasta la computadora que atiende esa dirección. |
| `Servidor -> Base de datos -> Servidor` | El servidor consulta los datos que necesita para responder. |
| `Servidor -> Navegador` | La respuesta regresa y el navegador la convierte en algo visible. |

# Módulo 1 — Entorno profesional y terminal

### A. Lo que vas a aprender
Vas a preparar tu "taller" de trabajo: el editor de código, la terminal y Git. Al terminar podrás crear una carpeta de proyecto y moverte por ella con confianza.

### B. ¿Qué es?
El editor de código (usaremos VS Code) es donde escribes y organizas tus archivos. La terminal es una forma de hablarle directamente a tu computadora escribiendo comandos, en vez de hacer clic en íconos. Git es un programa que registra los cambios de tu proyecto a lo largo del tiempo, y GitHub es un servicio en internet donde puedes guardar copias de esos proyectos.

### C. ¿Por qué existe?
Sin terminal, muchas herramientas del desarrollo web (instalar paquetes, ejecutar un servidor, usar Git) simplemente no funcionan igual de bien. Sin Git, perder un archivo o un cambio importante significa perderlo para siempre. Aprender esto primero evita frustraciones más adelante.

### D. ¿Cómo funciona? (paso a paso)
1. Abres la terminal integrada en VS Code (o la terminal del sistema).
2. Escribes un comando y presionas Enter.
3. La terminal ejecuta ese comando y muestra el resultado como texto.
4. Te mueves entre carpetas con comandos, sin usar el mouse.
5. Git observa la carpeta y puede registrar cualquier cambio que hagas en ella.

### E. Diagrama visual

```python
CARPETA DEL PROYECTO  (tu taller)
   |
   +-- VS Code            -> mesa de trabajo (donde escribes)
   +-- Terminal            -> panel de control (donde das ordenes)
   +-- Git (dentro de .git) -> maquina del tiempo (guarda versiones)
   +-- GitHub (en internet) -> copia remota del taller
```

### F. Ejemplo cotidiano (analogía)
> La carpeta del proyecto es tu taller. VS Code es tu mesa de trabajo. La terminal es el panel de control del taller. Git es una máquina del tiempo que guarda fotografías del taller en distintos momentos. GitHub es una bodega externa donde guardas copias de esas fotografías por si tu taller se incendia.

# Módulo 2 — HTML semántico y accesibilidad

### A. Lo que vas a aprender
Vas a construir la estructura de una página usando HTML, y a que esa estructura tenga significado, no solo apariencia. Al terminar podrás crear una tarjeta de evento completa.

### B. ¿Qué es?
HTML (HyperText Markup Language) describe qué es cada parte de una página: un título, un párrafo, una imagen, un botón. No decide colores ni distribución (eso es CSS); describe significado y estructura. Una etiqueta (tag) marca el inicio y el final de un elemento, y un atributo agrega información extra a ese elemento.

### C. ¿Por qué existe?
El navegador, los buscadores y las personas que usan lectores de pantalla necesitan saber qué es cada parte de la página para poder mostrarla, indexarla o leerla en voz alta correctamente. Un HTML con significado (semántico) hace todo eso posible; un HTML hecho solo de cajas genéricas (`div`) no comunica nada de eso.

### D. ¿Cómo funciona? (paso a paso)
1. El navegador lee el archivo HTML de arriba hacia abajo.
2. Cada etiqueta le indica al navegador qué tipo de contenido sigue.
3. El navegador construye una representación interna de la página (el DOM, que verás en el Módulo 5).
4. Esa representación se dibuja en pantalla siguiendo el orden del documento.

### E. Diagrama visual

```python
<main>                     ZONA PRINCIPAL
  <article>                UNA PIEZA DE CONTENIDO
    <h2>Titulo</h2>        ENCABEZADO
    <p>Texto...</p>        PARRAFO
    <button>Ver</button>   ACCION
  </article>
</main>
```

### F. Ejemplo cotidiano (analogía)
> Si una casa tuviera solo cajas sin etiquetas, sería difícil saber cuál es la cocina o el dormitorio. Las etiquetas semánticas de HTML son esas etiquetas de las cajas: `header`, `nav`, `main`, `section`, `article` y `footer` explican el papel de cada zona de la casa.

# Módulo 3 — CSS y responsive design

### A. Lo que vas a aprender
Vas a darle apariencia, orden y adaptabilidad a la interfaz. Al terminar podrás construir una cuadrícula de eventos que se adapta a distintos tamaños de pantalla.

### B. ¿Qué es?
CSS (Cascading Style Sheets) decide cómo se ven y se distribuyen los elementos HTML: colores, tamaños, espacios y posición. Un selector indica a qué elemento(s) aplica un estilo, una propiedad indica qué característica cambiar (color, tamaño...) y un valor indica cómo cambiarla.

### C. ¿Por qué existe?
Separar estructura (HTML) de apariencia (CSS) permite cambiar el diseño completo de una página sin tocar su contenido, y reutilizar el mismo diseño en muchas páginas. Sin CSS, cada página web se vería con la tipografía y el espaciado por defecto del navegador.

### D. ¿Cómo funciona? (paso a paso)
1. El navegador lee el HTML y construye la estructura de la página.
2. El navegador lee el CSS y calcula qué estilos aplican a cada elemento.
3. Cada elemento se dibuja como una caja rectangular (box model): contenido, padding, borde y margen.
4. Flexbox y Grid deciden cómo se organizan esas cajas entre sí.
5. Las media queries cambian esas reglas según el tamaño de la pantalla.

# Módulo 4 — JavaScript esencial

### A. Lo que vas a aprender
Vas a aprender a expresar reglas y resolver problemas con código. Al terminar podrás escribir una función que calcule un descuento según reglas.

### B. ¿Qué es?
JavaScript es el lenguaje que permite que la página piense y reaccione. Una variable guarda un valor (un número, un texto...). Una condición decide entre caminos distintos según se cumpla o no algo. Un bucle repite una acción varias veces. Una función agrupa una acción que puede reutilizarse las veces que quieras.

### C. ¿Por qué existe?
HTML y CSS son estáticos: muestran siempre lo mismo. JavaScript agrega lógica: calcular precios, validar formularios, decidir qué mostrar según una condición. Sin JavaScript, una página no puede reaccionar a lo que hace la persona.

### D. ¿Cómo funciona? (paso a paso)
1. El navegador ejecuta el código JavaScript en orden, de arriba hacia abajo.
2. Las variables guardan valores en la memoria mientras el programa corre.
3. Cuando el programa llega a un `if`, evalúa la condición y sigue por una rama u otra.
4. Cuando el programa llama a una función, "salta" a ejecutar ese bloque y regresa con un resultado.

# Módulo 5 — DOM, eventos y formularios

### A. Lo que vas a aprender
Vas a conectar JavaScript con lo que el usuario ve y hace. Al terminar podrás agregar una tarjeta nueva a la página sin recargarla.

### B. ¿Qué es?
El DOM (Document Object Model) es la representación de la página que el navegador mantiene en memoria mientras la muestra. JavaScript puede buscar elementos dentro de esa representación, leer su contenido y modificarlo. Un evento es algo que ocurre en la página: un clic, escribir en un campo, enviar un formulario, y JavaScript puede "escuchar" esos eventos y reaccionar.

### C. ¿Por qué existe?
Sin el DOM, JavaScript no tendría forma de "tocar" lo que ya existe en la página. Sin eventos, JavaScript no sabría cuándo actuar: necesita saber que el usuario hizo clic en un botón exacto, en un momento exacto, para reaccionar ahí.

### D. ¿Cómo funciona? (paso a paso)
1. El navegador construye el DOM a partir del HTML al cargar la página.
2. JavaScript usa `document.querySelector` (u otros métodos) para encontrar un elemento del DOM.
3. JavaScript registra un "escucha" (`addEventListener`) sobre ese elemento para un tipo de evento.
4. Cuando el evento ocurre, el navegador ejecuta la función asociada.
5. Esa función puede leer datos, cambiar texto, agregar o quitar elementos del DOM.

### E. Diagrama visual

```python
HTML (plano)  --navegador construye-->  DOM (maqueta viva en memoria)
                                              |
                                    JavaScript busca un elemento
                                              |
                                    addEventListener('click', fn)
                                              |
                                     usuario hace click
                                              |
                                       se ejecuta fn()
                                              |
                                    el DOM cambia -> la pantalla cambia
```

### F. Ejemplo cotidiano (analogía)
> El HTML es el plano de una casa; el DOM es la maqueta viva de esa casa que existe en memoria mientras la visitas. JavaScript puede entrar a esa maqueta y mover piezas (cambiar un texto, agregar un mueble) cada vez que ocurre un evento, como tocar el timbre.

### A. Lo que vas a aprender

### B. ¿Qué es?
Git no es lo mismo que GitHub. Git vive dentro de tu proyecto y registra cambios con el tiempo. GitHub es un servicio en internet donde puedes guardar ese repositorio y compartirlo con otras personas. Un commit es una fotografía de tu proyecto en un momento dado, acompañada de un mensaje que explica qué cambió.

### C. ¿Por qué existe?
Sin Git, si rompes algo que funcionaba, no hay forma sencilla de volver atrás. Git te permite experimentar con confianza, porque siempre puedes regresar a una versión anterior que sabías que funcionaba. GitHub además permite compartir y respaldar tu trabajo fuera de tu computadora.

### D. ¿Cómo funciona? (paso a paso)
1. Modificas archivos dentro de tu proyecto.
2. Le dices a Git cuáles cambios quieres guardar (`git add`).
3. Creas una fotografía de esos cambios con un mensaje (`git commit`).
4. Git guarda esa fotografía en el historial del proyecto.
5. Cuando quieres, envías ese historial a GitHub (`git push`) para respaldarlo o compartirlo.

# Módulo 7 — HTTP, JSON y APIs

### A. Lo que vas a aprender
Vas a entender cómo viajan las solicitudes y los datos entre el navegador y un servidor. Al terminar podrás diseñar en papel una API completa.

### B. ¿Qué es?
HTTP es un protocolo de comunicación: un conjunto de reglas que navegadores y servidores usan para entenderse. Una solicitud (request) indica qué se quiere hacer, sobre qué recurso y con qué datos. Una respuesta (response) incluye un estado, encabezados y normalmente un cuerpo con información. JSON es un formato de texto muy común para transportar esos datos. Un endpoint es una dirección específica de una API diseñada para una operación concreta.

### C. ¿Por qué existe?
El navegador y el servidor son programas distintos, quizás en computadoras distintas. Necesitan un lenguaje común y predecible para pedir y entregar información: eso es HTTP. Una API es la "puerta de entrada" que un backend expone para que otros programas (como tu frontend) puedan pedirle cosas de forma ordenada.

### D. ¿Cómo funciona? (paso a paso)
1. El cliente (navegador) arma una solicitud: método, dirección, y a veces datos.
2. La solicitud viaja por internet hasta el servidor correcto.
3. El servidor identifica el endpoint solicitado y ejecuta la lógica correspondiente.
4. El servidor arma una respuesta con un código de estado y, normalmente, datos en JSON.
5. El cliente recibe la respuesta y decide qué hacer con ella (mostrarla, guardarla, etc.).

### E. Diagrama visual

```python
CLIENTE                          SERVIDOR
   |                                  |
   |  GET /eventos                    |
   |--------------------------------->|
   |                                  | procesa
   |     200 OK  [ {...}, {...} ]     |
   |<---------------------------------|
   |                                  |
```

### F. Ejemplo cotidiano (analogía)

# Módulo 8 — Asincronía y fetch

### A. Lo que vas a aprender
Vas a aprender por qué el navegador puede seguir funcionando mientras espera datos de internet. Al terminar podrás consumir una API real desde JavaScript.

### B. ¿Qué es?
Algunas tareas tardan: pedir datos a internet, leer un archivo, consultar un servidor. JavaScript puede esperar esos resultados sin congelar toda la interfaz gracias a la asincronía. `fetch` es una función que permite hacer solicitudes HTTP desde el navegador. Una promesa (promise) representa un valor que estará disponible en el futuro; `async` y `await` son una forma más legible de trabajar con promesas.

### C. ¿Por qué existe?
Si el navegador se detuviera por completo cada vez que pide datos a internet, la página quedaría congelada mientras espera, y eso sería una pésima experiencia. La asincronía permite seguir interactuando con la página mientras la respuesta viaja de vuelta.

### D. ¿Cómo funciona? (paso a paso)
1. El código llama a `fetch(url)`, que empieza la solicitud pero no bloquea el resto del programa.
2. Mientras se espera la respuesta, el navegador sigue funcionando con normalidad.
3. Cuando la respuesta llega, el código que sigue después de `await` se ejecuta con el resultado.
4. Si algo falla (sin conexión, error del servidor), el bloque `catch` se encarga de manejarlo.

### E. Diagrama visual

```python
   loadEvents() inicia
        |
        v
   fetch('/api/eventos')  ---> (esperando... la pagina sigue viva)
        |
        v
   llega la respuesta
        |
    +---+---+
    |       |
   ok      error
    |       |
   json()  catch(error)
```

### F. Ejemplo cotidiano (analogía)
> No necesitas quedarte parado mirando la puerta mientras esperas a un repartidor. Puedes seguir con otras cosas y recibir una notificación cuando llegue. `await` es ese "seguir con lo mío, pero avísame cuando llegue".

### G. Ejemplo dentro del proyecto EventHub
Harás que EventHub pida la lista real de eventos a una API (por ahora pública, de práctica) y la muestre en pantalla, con un mensaje de carga y uno de error.

### H. Código mínimo

Lee primero el código y trata de predecir el resultado. Después modifica al menos una parte y vuelve a probar.

```python
async function loadEvents() {
  try {
    const response = await fetch('/api/eventos');
    if (!response.ok) throw new Error('Error HTTP');
    return await response.json();
  } catch (error) {
    console.error(error);
  }
}
```

### I. Explicación línea por línea

| Línea | Qué hace |
|---|---|
| `async function loadEvents() {` | Declara una función asíncrona: puede usar `await` dentro de ella. |
| `try {` | Inicia un bloque donde vigilamos si ocurre algún error. |
| `const response = await fetch('/api/eventos');` | Pide los datos y espera la respuesta antes de continuar. |
| `if (!response.ok) throw new Error('Error HTTP');` | Si la respuesta indica un error (por ejemplo 404 o 500), lanza un error manualmente. |
| `return await response.json();` | Convierte el cuerpo de la respuesta de texto a un objeto/array de JavaScript y lo devuelve. |
| `} catch (error) {` | Si algo falló en el bloque try, este código se ejecuta con el error capturado. |
| `console.error(error);` | Muestra el error en la consola para poder investigarlo. |

### J. ¿Qué debería ver en pantalla?
Si la API responde correctamente, la función devuelve un array de eventos; si hay un problema de red o el servidor falla, verás el error impreso en la consola en vez de que la página se rompa en silencio.

### K. Error frecuente
⚠️ Olvidar manejar errores de red, asumiendo que `fetch` siempre va a funcionar, lo que hace que la aplicación se rompa sin explicación cuando el servidor falla o no hay conexión.

### L. Cómo encontrar el error

# Módulo 9 — Arquitectura de frontend

### A. Lo que vas a aprender
Vas a aprender a separar responsabilidades para que la interfaz pueda crecer sin volverse un caos. Al terminar podrás dividir tu código en funciones que hacen una sola cosa cada una.

### B. ¿Qué es?
Cuando una app crece, conviene separar datos, presentación y lógica en piezas distintas. Una función que solo pide datos, una que solo los transforma y una que solo los muestra son más fáciles de entender, probar y corregir que una única función gigante que hace todo a la vez.

### C. ¿Por qué existe?
Sin separación, cualquier cambio pequeño (por ejemplo, cambiar de dónde vienen los datos) obliga a tocar código que también dibuja la pantalla, y es fácil romper algo sin darse cuenta. Separar responsabilidades reduce el acoplamiento: cada pieza puede cambiar sola.

### D. ¿Cómo funciona? (paso a paso)
1. Una función se encarga solo de pedir datos (por ejemplo, a una API).
2. Otra función, si hace falta, transforma esos datos (filtra, ordena, calcula).
3. Otra función se encarga solo de dibujar esos datos en el DOM.
4. El código que las conecta es corto y solo coordina el orden en que se llaman.

### E. Diagrama visual

```python
   getEvents()  ------->  datos crudos
        |
        v
   filterEvents(datos)  ------->  datos filtrados
        |
        v
   renderEvents(datos)  ------->  DOM actualizado
```

### F. Ejemplo cotidiano (analogía)
> Una cocina profesional separa almacén, preparación, cocción y salida. Si todo ocurriera en una sola mesa, con una sola persona, cualquier cambio en el menú sería muy difícil de coordinar.

### A. Lo que vas a aprender

### B. ¿Qué es?
Node.js es un entorno de ejecución (runtime) de JavaScript que permite correr ese lenguaje fuera del navegador: en tu computadora o en un servidor. Gracias a Node.js, el código del backend recibe solicitudes, valida datos, ejecuta reglas de negocio y devuelve respuestas.

El navegador ejecuta JavaScript pensado para interfaces: botones, formularios, el DOM. Pero un servidor necesita hacer otras cosas: escuchar peticiones de red, leer archivos, hablar con una base de datos. Node.js permite usar el mismo lenguaje (JavaScript) también para eso.

### D. ¿Cómo funciona? (paso a paso)

# Módulo 11 — Express y REST API

### A. Lo que vas a aprender
Vas a construir endpoints de una API de forma organizada. Al terminar tendrás una API de eventos completa con las cuatro operaciones básicas.

### B. ¿Qué es?
Express es una herramienta (framework) que facilita crear rutas y middleware en Node.js, en vez de escribir todo a mano como en el módulo anterior. Una API REST usa recursos (como "eventos") y métodos HTTP para expresar operaciones comunes. CRUD significa Crear, leer (Read), actualizar (Update) y eliminar (Delete): las cuatro operaciones más comunes sobre cualquier recurso.

### C. ¿Por qué existe?
Escribir un servidor a mano para cada ruta, con validaciones y manejo de errores, se vuelve repetitivo y propenso a errores. Express organiza ese trabajo con una sintaxis clara para definir rutas, y con middleware: funciones que se ejecutan antes de llegar a la lógica final (por ejemplo, para leer el cuerpo de la solicitud).

### D. ¿Cómo funciona? (paso a paso)
1. Se crea una aplicación Express (`express()`).
2. Se definen rutas: un método HTTP, una dirección y una función que responde.
3. El middleware (como `express.json()`) procesa la solicitud antes de que llegue a la ruta.
4. Cada ruta responde con datos (normalmente JSON) o con un código de error si algo falla.

### E. Diagrama visual

```python
NAVEGADOR
   |
   |  GET /eventos
   v
EXPRESS  ---busca la ruta que coincide--->  app.get('/eventos', funcion)
   |
   v
FUNCION se ejecuta
   |
   v
res.json([...])  ---> JSON de vuelta al navegador
```

### F. Ejemplo cotidiano (analogía)
> Las rutas de Express son como mostradores especializados: cada uno recibe un tipo específico de solicitud ("quiero ver la lista", "quiero crear uno nuevo") y entrega una respuesta siguiendo reglas claras.

### G. Ejemplo dentro del proyecto EventHub
Construirás la API real de EventHub con Express: listar eventos, ver uno por id, crear, actualizar y eliminar, devolviendo errores claros cuando el recurso no existe.

### H. Código mínimo

Lee primero el código y trata de predecir el resultado. Después modifica al menos una parte y vuelve a probar.

```python
const express = require('express');
const app = express();
app.use(express.json());

app.get('/eventos', (req, res) => {
  res.json([{ id: 1, nombre: 'Festival' }]);
});

app.listen(3000);
```

### I. Explicación línea por línea

| Línea | Qué hace |

# Módulo 12 — SQL y PostgreSQL

### A. Lo que vas a aprender
Vas a aprender a almacenar y consultar datos de forma estructurada. Al terminar podrás crear tablas relacionadas y hacer consultas básicas.

### B. ¿Qué es?
Una base de datos relacional organiza la información en tablas relacionadas entre sí. SQL (Structured Query Language) es el lenguaje para consultar y modificar esos datos. PostgreSQL es un sistema de gestión de bases de datos relacional, uno de los más usados en el mundo real. Una tabla tiene filas (registros) y columnas (campos). La clave primaria (primary key) identifica de forma única cada fila; una clave foránea (foreign key) conecta una tabla con otra.

### C. ¿Por qué existe?
Guardar datos en variables o archivos sueltos funciona para pruebas pequeñas, pero no resiste miles de registros, múltiples usuarios accediendo a la vez, ni relaciones complejas entre datos. Las bases de datos relacionales están diseñadas justamente para eso: reglas, relaciones, concurrencia y consultas a mayor escala.

### D. ¿Cómo funciona? (paso a paso)
1. Se define una tabla con columnas y sus tipos de dato (texto, número, fecha...).
2. Cada fila nueva se guarda de forma permanente en la base de datos.
3. SQL permite pedir (SELECT), insertar (INSERT), modificar (UPDATE) o borrar (DELETE) esas filas.
4. Las relaciones entre tablas (por ejemplo, eventos y usuarios) se hacen con claves foráneas.

# Módulo 13 — Persistencia: Express + PostgreSQL

### A. Lo que vas a aprender
Vas a conectar tu API con datos permanentes. Al terminar, los datos de EventHub sobrevivirán aunque reinicies el servidor.

### B. ¿Qué es?
Hasta ahora el servidor puede responder, pero los datos escritos directamente en el código desaparecen al reiniciarlo. Aquí conectamos Express con PostgreSQL para que las rutas lean y escriban datos reales y permanentes. Una consulta parametrizada es una forma segura de insertar valores dentro de una consulta SQL sin concatenar texto directamente.

### C. ¿Por qué existe?
Una aplicación real necesita que los datos sobrevivan reinicios, fallos y el paso del tiempo. Guardar todo en variables de JavaScript es útil para practicar, pero no sirve para un producto real; PostgreSQL resuelve ese problema de forma robusta.

### D. ¿Cómo funciona? (paso a paso)
1. El servidor abre una conexión (o un pool de conexiones) hacia la base de datos.
2. Cuando llega una solicitud, la ruta ejecuta una consulta SQL usando esa conexión.
3. Los valores que vienen del usuario se pasan como parámetros, nunca pegados directamente al texto SQL.
4. El resultado de la consulta se transforma en la respuesta que recibe el cliente.

### E. Diagrama visual

```python
NAVEGADOR --GET /eventos/15--> EXPRESS --query SQL--> POSTGRESQL
                                                             |
NAVEGADOR <--------- JSON <----- EXPRESS <---- filas -------+
```

### F. Ejemplo cotidiano (analogía)
> El backend es el encargado de tomar el pedido; PostgreSQL es el registro permanente donde se conserva la información, como un archivero que no se borra aunque apagues las luces de la oficina.

# Módulo 14 — Autenticación y autorización

### A. Lo que vas a aprender
Vas a distinguir identidad de permisos. Al terminar podrás proteger una ruta exclusiva para administradores.

### B. ¿Qué es?
Autenticación responde "¿quién eres?". Autorización responde "¿qué puedes hacer?". Nunca se deben guardar contraseñas en texto plano: deben almacenarse con funciones de hash adecuadas. Un token o una sesión es la forma en que el servidor recuerda que ya verificó tu identidad en solicitudes futuras.

### C. ¿Por qué existe?
Muchas aplicaciones necesitan saber quién hace cada solicitud (para mostrar datos personalizados, proteger información) y qué le está permitido hacer (un usuario normal no debería poder borrar eventos ajenos). Separar estos dos conceptos evita errores graves de seguridad.

### D. ¿Cómo funciona? (paso a paso)
1. El usuario se registra: su contraseña se transforma con hash antes de guardarse.
2. El usuario inicia sesión: el servidor compara el hash de la contraseña enviada con el guardado.
3. Si coincide, el servidor entrega una prueba de identidad (sesión o token) al cliente.
4. En solicitudes futuras, el cliente envía esa prueba, y el servidor decide qué le permite hacer según su rol.

### E. Diagrama visual

```python
REGISTRO
   |
   v
USUARIO (contraseña guardada como hash)
   |
   v
LOGIN  --credenciales correctas-->  SESION / TOKEN
   |
   v
RECURSOS PROTEGIDOS  (segun el rol del usuario)
```

### F. Ejemplo cotidiano (analogía)
> Al entrar a un edificio, primero verifican tu identidad (autenticación) en la recepción. Después, una tarjeta de acceso decide a qué habitaciones puedes entrar (autorización): estar identificado no significa tener acceso a todo.

# Módulo 15 — Seguridad web básica

### A. Lo que vas a aprender
Vas a reconocer riesgos comunes y aprender a reducir su impacto. Al terminar podrás identificar y corregir vulnerabilidades típicas en una API.

### B. ¿Qué es?
La seguridad no es un módulo que se agrega al final: es una forma de pensar durante todo el desarrollo. Implica validar entradas, limitar permisos, proteger secretos (contraseñas, claves) y nunca confiar ciegamente en datos enviados por el cliente.

### C. ¿Por qué existe?
Cualquier aplicación conectada a internet puede recibir solicitudes maliciosas, no solo de usuarios legítimos. Sin prácticas básicas de seguridad, datos sensibles pueden filtrarse, modificarse o borrarse por personas no autorizadas.

### D. ¿Cómo funciona? (paso a paso)
1. Cada amenaza tiene un patrón: alguien envía datos manipulados esperando que el sistema los use sin revisar.
2. La defensa suele consistir en validar, escapar o parametrizar esos datos antes de usarlos.
3. Los secretos (contraseñas de base de datos, claves de API) se guardan fuera del código, en variables de entorno.
4. CORS controla qué otros sitios pueden hacer solicitudes a tu API desde el navegador.

### E. Diagrama visual

### F. Ejemplo cotidiano (analogía)
> Una puerta con buena cerradura sigue siendo insegura si dejas las llaves pegadas en la puerta. La seguridad implica pensar tanto en el acceso (la cerradura) como en el comportamiento posterior (dónde dejas las llaves).

# Módulo 16 — Testing y debugging

### A. Lo que vas a aprender
Vas a aprender a encontrar errores de manera sistemática y a demostrar que tu software funciona. Al terminar podrás escribir tu primera prueba automatizada.

### B. ¿Qué es?
Debugging es investigar por qué el programa no se comporta como esperabas. Testing es comprobar de manera repetible que ciertas reglas siguen funcionando, incluso después de cambiar el código. Los dos son habilidades normales del desarrollo, no un castigo posterior a programar.

### C. ¿Por qué existe?
El código cambia constantemente, y sin pruebas, es fácil romper algo que funcionaba sin darte cuenta (una regresión). El debugging sistemático evita perder horas "adivinando" la causa de un error.

### D. ¿Cómo funciona? (paso a paso)
1. Debugging: reproducir el error, observar evidencia (consola, Network), aislar la causa, probar una hipótesis, corregir y verificar.
2. Testing: escribir una entrada conocida, ejecutar la función, y comprobar si el resultado es el esperado.
3. Un test automatizado se puede volver a correr en segundos cada vez que cambias el código.

### E. Diagrama visual

```python
TESTING:
  ENTRADA --> FUNCION --> RESULTADO --> ¿ES LO ESPERADO?

DEBUGGING:
  ¿Que esperaba?
       |
  ¿Que ocurrio?
       |
  ¿Donde cambio el comportamiento?
       |
  ¿Que hipotesis tengo?
       |
  ¿Como la pruebo?
       |
  ¿Funciono?
```

### F. Ejemplo cotidiano (analogía)
> Debugging es trabajo detectivesco: buscas pistas y sospechosos hasta encontrar al culpable. Testing es como una lista de chequeo que repites cada vez que algo cambia, para confirmar que nada se rompió.

# Módulo 17 — Frontend moderno con React

### A. Lo que vas a aprender
Vas a entender qué problema resuelve un framework como React y a construir componentes reutilizables. Al terminar podrás convertir una tarjeta de evento en un componente.

### B. ¿Qué es?
Nuestro proyecto está creciendo. Tenemos tarjetas repetidas, formularios, listas que cambian según filtros: estados que cambian constantemente. React ayuda a construir interfaces a partir de componentes (piezas reutilizables de interfaz) y estado (los datos que pueden cambiar dentro de la aplicación).

### C. ¿Por qué existe?
Manipular el DOM directamente, como en el Módulo 5, funciona bien para páginas pequeñas, pero se vuelve difícil de mantener cuando hay muchas piezas interconectadas que cambian. React organiza esa complejidad: describes cómo se ve la interfaz según los datos actuales, y React se encarga de actualizar el DOM quwe cambió.

### D. ¿Cómo funciona? (paso a paso)
1. Divides la interfaz en componentes: piezas reutilizables con su propia lógica y apariencia.
2. Cada componente puede recibir datos desde afuera (props) y tener sus propios datos internos que cambian (state).
3. Cuando el estado cambia, React vuelve a calcular qué debe mostrarse y actualiza solo lo necesario en el DOM.
4. Los hooks (como useState) son funciones especiales que dan estas capacidades a los componentes.

### E. Diagrama visual

```python
ANTES (Modulo 5):           DESPUES (React):

HTML fijo                   <EventList>
   |                              |
JS busca y modifica            +--+--+
el DOM a mano                  |     |
                          <EventCard> <EventCard>
                           (props: un evento cada una)
```

### F. Ejemplo cotidiano (analogía)
> Si tu interfaz fuera una ciudad, un componente es un edificio reutilizable (puedes construir el mismo diseño de edificio en varios lugares) y el estado son los datos que cambian dentro de esa ciudad, como el número de personas en cada edificio en un momento dado.

### A. Lo que vas a aprender

### B. ¿Qué es?
Deployment significa publicar la aplicación en una infraestructura real, fuera de tu computadora. Aquí aparecen conceptos como build (preparar el código para producción), variables de entorno, dominio, HTTPS y logs de producción.

### C. ¿Por qué existe?
Una aplicación que solo funciona en tu computadora no sirve de mucho para otras personas. El deployment es el paso que convierte tu proyecto de práctica en algo que cualquiera puede visitar desde un navegador, en cualquier parte del mundo.

### D. ¿Cómo funciona? (paso a paso)

5. El servicio te entrega una dirección (dominio) donde tu app queda disponible con HTTPS.

# Qué hacer cuando algo no funciona

Guía general de debugging que aplica a cualquier módulo del curso.

**1. Reproduce:** Haz que el error vuelva a ocurrir de forma consistente.
**2. Observa:** Mira consola, Network, status code y datos reales.
**3. Aísla:** Reduce el problema a la pieza más pequeña posible.
**4. Formula una hipótesis:** Escribe qué crees que está pasando y por qué.
**5. Prueba:** Cambia una sola cosa y observa el efecto.
**6. Corrige:** Haz el cambio mínimo que resuelve la causa.
**7. Verifica:** Vuelve a ejecutar el caso original y una variante.
**8. Documenta:** Registra la causa y la solución si fue un error importante.

> **Frase clave: no adivines. Observa evidencia.**

```python
# Ejercicio mental de debugging
expected = 300
actual = 100 * 2
print('expected:', expected)
print('actual:', actual)
print('¿Coinciden?', expected == actual)
```

# Glosario para principiantes

**API**  
Interfaz con reglas que permite que un programa solicite o envíe datos a otro programa.

**Backend**  
Código que corre en el servidor y procesa peticiones, reglas y datos.

**Base de datos**  
Sistema que almacena información de forma estructurada y permite consultarla.

**Branch**  
Línea de trabajo independiente en Git para desarrollar cambios sin afectar la rama principal.

**Commit**  
Registro de un conjunto de cambios en Git.

**DOM**  
Representación de la página que el navegador mantiene en memoria para poder manipularla con JavaScript.

**Endpoint**  
Ruta específica de una API que recibe solicitudes y produce respuestas.

**Frontend**  
Parte de la aplicación que el usuario ve y con la que interactúa.

**HTTP**  
Protocolo que define cómo clientes y servidores intercambian solicitudes y respuestas en la web.

**JSON**  
Formato de texto común para representar objetos y listas de datos.

**Middleware**  
Función que procesa una solicitud antes de que llegue a la lógica final de una ruta.

**Node.js**  
Entorno que permite ejecutar JavaScript fuera del navegador.

**Repositorio**  
Carpeta gestionada por Git que contiene archivos y el historial de cambios.

**Responsive**  
Diseño que se adapta a distintos tamaños y formas de pantalla.

**REST**  
Estilo de diseño de APIs basado en recursos y operaciones HTTP coherentes.

**Runtime**  
Entorno que proporciona lo necesario para ejecutar un programa.

**Servidor**  
Sistema que recibe solicitudes y devuelve respuestas o servicios a clientes.

**SQL**  
Lenguaje utilizado para consultar y modificar bases de datos relacionales.

**UI**  
Interfaz de usuario: lo que una persona ve y utiliza.

**UX**  
Experiencia de usuario: facilidad, claridad y calidad con que una persona logra un objetivo.
